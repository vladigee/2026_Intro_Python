## Exercise 1: A year of station data

The practice exercises used values typed by hand. This one uses a real record: daily maximum
temperatures for 2022 from the automated weather station at Independence Municipal Airport,
Iowa, US, station code IIB. The file has one row for each of the 365 days, but 16 of them, from
25 October to 9 November, have no temperature value. Work through the steps in order, since each
step reuses variables from the one before. The same file returns in 1.2, where you turn the whole
year into a monthly summary.

The data has three columns: the station code, the day as `dd.mm.yy`, and the daily maximum
temperature **in degrees Fahrenheit**. Converting to SI units is part of the job.

```python
# Pre-supplied: download the data file and cache it locally.
# You do not need to understand this cell yet — fetching data is covered in the
# reproducible-data-pipelines bonus subchapter.
import pooch

data_file = pooch.retrieve(
    url="https://raw.githubusercontent.com/gse-unil/2026_MLEES_book/main/data/part-I/station_iib_daily_max_temp_2022.csv",
    known_hash="sha256:034755fb289b4e157e5d029995481786a359eb25a80df62c09ee83d063013144",
    fname="station_iib_daily_max_temp_2022.csv",
    path=pooch.os_cache("mlees"),
)
print("data cached at:", data_file)
```

**Step 1.** Open the file with a `with` block in read mode and read all its lines with
`.readlines()`.

Print how many lines the file has, then print the first three lines exactly as
they come off disk.

```python
# Your solution here
# Hint: data_file is a plain string; after from pathlib import Path, Path(data_file) is a
#       path you can .open()
# Hint: with Path(data_file).open("r", encoding="utf-8") as fhandler: ...
# Hint: .readlines() gives you a list of strings, one per line
# Hint: lines[0] is the header row, not an observation
# Hint: print lines[0], lines[1] and lines[2] one at a time

#data_file = r"C:\Users\redqu\AppData\Local\mlees\mlees\Cache\station_iib_daily_max_temp_2022.csv"

with Path(data_file).open("r", encoding="utf-8") as fhandler:
    read_all_lines = fhandler.readlines()
    #print(read_all_lines)
    print(len(read_all_lines))
    #print(read_all_lines[0:3])
    print("".join(read_all_lines[0:3]), end="")
```

**Step 2.** Look at the first record (the second line of the file). Split it on the separator
and print the resulting list. Then print the *type* of each of the three fields.

Answer in a comment: are the temperatures numbers at this point?

```python
# Your solution here
# Hint: .strip() first, to remove the trailing newline
# Hint: .split(",") turns one line into a list of fields
# Hint: type() will tell you what you are really holding

stripped = read_all_lines[1].strip()
splitted = stripped.split(',')

print(splitted)

print(type(splitted[0]))
print(type(splitted[1]))
print(type(splitted[2]))

# temperature numbers are strings
```

**Step 3.** Take that same first record and turn it into three properly typed, well-named
variables: the station code as text, the day as text, and the maximum temperature as a
**float in degrees Celsius**. Print them in one line with an f-string, showing the temperature
to one decimal with its unit.

Recall that °C = (°F − 32) × 5/9.

```python
# Your solution here
# Hint: name each variable for the quantity and unit it holds, e.g. day_1 and
#       max_temp_celsius_1; step 4 needs a second set of names for the second day
# Hint: f"{value:.1f} °C" formats to one decimal

station_code_1 = splitted[0]
day_1 = splitted[1]
max_temp_celsius_1_str = splitted[2]
max_temp_celsius_1 = float(max_temp_celsius_1_str)
max_temp_celsius_1 = (max_temp_celsius_1-32) * 5/9

print(f"Station code: {station_code_1}, Day: {day_1}, Max Temperatur: {max_temp_celsius_1:.1f} °C")
```

**Step 4.** Now do the same for the *second* record (the third line of the file), so you have
two days side by side.

1. Ask which of the two days was warmer by comparing the two raw temperature **fields**, before
   any casting.
2. Ask the same question again, this time comparing the two values **after** casting to float.
3. The two answers disagree. In a comment, explain which one is right and why the other one is
   wrong.

```python
# Your solution here
# Hint: you already have the first record's fields; do the same for lines[2]
# Hint: compare the raw strings first, then compare float() of each
# Hint: give the second day its own names, e.g. day_2 and max_temp_celsius_2
stripped = read_all_lines[2].strip()
splitted = stripped.split(',')
print(splitted)

station_code_2 = splitted[0]
day_2 = splitted[1]
max_temp_celsius_2_str = splitted[2]
max_temp_celsius_2 = float(max_temp_celsius_2_str)
max_temp_celsius_2 = (max_temp_celsius_2-32) * 5/9

print(f"Station code: {station_code_2}, Day: {day_2}, Max Temperatur: {max_temp_celsius_2:.1f} °C")

# 1
day_1_warmer = max_temp_celsius_1_str > max_temp_celsius_2_str
day_2_warmer = max_temp_celsius_2_str > max_temp_celsius_1_str
print(f'Is Day 1 warmer than Day 2? {day_1_warmer}. Is Day 2 warmer than Day 1? {day_2_warmer}.')

# or

if max_temp_celsius_1_str > max_temp_celsius_2_str:
    print("Day 1 is warmer.")
elif max_temp_celsius_2_str > max_temp_celsius_1_str:
    print("Day 2 is warmer.")
else:
    print("Both days are the same temperature.")

# or

print(max_temp_celsius_1_str > max_temp_celsius_2_str)

# 2
day_1_warmer_float = max_temp_celsius_1 > max_temp_celsius_2
day_2_warmer_float = max_temp_celsius_2 > max_temp_celsius_1
print(f'Is Day 1 warmer than Day 2? {day_1_warmer_float}. Is Day 2 warmer than Day 1? {day_2_warmer_float}.')

# or

if max_temp_celsius_1 > max_temp_celsius_2:
    print("Day 1 is warmer.")
elif max_temp_celsius_2 > max_temp_celsius_1:
    print("Day 2 is warmer.")
else:
    print("Both days are the same temperature.")

# or

print(max_temp_celsius_1 > max_temp_celsius_2)

# 3: the one with the float is the correct answer, the other one compares the characters of the strings.
```
