We’ve recorded the daily temperatures in Lausanne for the month of December, and saved them in a text file.
Each line of the file contains one temperature in °C with 2 decimal places.

Write a Python function called average_temperature(filename) that:

Opens the file. Your function should work for any valid file name. 

Reads all the temperatures from the file and stores them in a list.

Computes the average temperature.

Returns the average temperature rounded to 1 decimal place.

Hint 1: You can use with open(filename) as f: to open the file and the method strip() to clean up each line of the file.

Hint 2: The file may contain several lines, so don't forget to loop through them. 

Par exemple:

Test	Résultat
print(average_temperature("temps0.txt"))
2.0
Réponse :(régime de pénalités : 10, 20, ... %)

```python

def average_temperature(filename):
    with open(filename, "r", encoding="utf-8") as f:
        read_data = f.readlines()
        list_temp = []
    for r in read_data:
        cleaned = r.strip()
        converted = float(r)
        list_temp.append(converted)
    avg_temp = sum(list_temp)/len(list_temp)
    avg_temp = round(avg_temp, 1)
    return avg_temp
```

We have rainfall data from weather stations stored in a Python dictionary. Each key is a station name (string) and each value is the rainfall in millimetres (float). Here's an example dictionary with rainfall data:

rainfall_dict = {
    'Station A': 45.2,
    'Station B': 67.5,
    'Station C': 23.1,
    'Station D': 88.0
}

Note that the dictionary may contain any number of stations with varying names. Your code should work correctly regardless of how many entries it has or what the station names are.

Write a Python function called check_rainfall(rainfall_dict) that:

Takes one argument rainfall_dict (a dictionary with rainfall data), and returns a list of strings.

Loops through the dictionary.

For each station:

If the rainfall is greater than 50 mm, add a string "StationName: Heavy rainfall" to a results list.

Otherwise add "StationName: Normal rainfall" to the results list.

Be careful to print the string with the right number of spaces (there's a space between the ":" and the "N"/"H". 
Returns the list of strings after processing all stations.

For example, for the above dictionary, the expected output is:
['Station A: Normal rainfall',
 'Station B: Heavy rainfall',
 'Station C: Normal rainfall',
 'Station D: Heavy rainfall']

```python
def check_rainfall(rainfall_dict):
    list_of_strings = []
    for s, r in rainfall_dict.items():
        if r > 50:
            list_of_strings.append(f"{s}: Heavy rainfall")
        else: list_of_strings.append(f"{s}: Normal rainfall")
    return list_of_strings
```

Your task is to create a function called compute_array_statistics that performs the following operations using the NumPy library:

The function should accept one argument, a Python list or NumPy array of numbers.

Inside the function, convert the input into a NumPy array.

Compute the mean of the CO_2 measurements using a built-in NumPy function.

Compute the standard deviation of the CO_2 measurements using a built-in NumPy function.

The function should return a tuple containing the mean and the standard deviation, in that order.

Use the following list for your measurements: co2_measurements = [421.5, 422.1, 420.8, 421.9, 423.0, 421.7, 422.5]. Note: the function does not take into account the edge case of an empty list.

Hint: To make your function return two variables, the syntax is return X, Y

```python
import numpy as np

def compute_array_statistics(list_or_array):
    convert = np.array(list_or_array)
    mean = np.mean(convert)
    std = np.std(convert)
    #tuple_mean_std = 
    #mean, std = tuple_mean_std
    return mean, std
```

Your task is to import the relevant libraries and create a function called analyze_air_quality that performs the following operations using the pandas library:

The function should accept one argument, a string containing the CSV data.

Inside the function, read the provided CSV data using io.StringIO() into a pandas DataFrame.

The function should return the DataFrame.

```python
import pandas as pd
import io
#print(help(io.StringIO))
def analyze_air_quality(string_csv):
    # io.StringIO is used to treat the string as a file-like object
    # so that you can use pandas functions as if you were reading a file.
    df = pd.read_csv(io.StringIO(string_csv))
    return df
```
